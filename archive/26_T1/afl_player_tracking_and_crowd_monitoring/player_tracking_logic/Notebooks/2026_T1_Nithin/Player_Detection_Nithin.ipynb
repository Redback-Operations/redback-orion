### The Split Script

```python
# =====================================================================
# Train/Val Split — Sprint 3 improvement
# Creates a reproducible 80/20 split from the original 200 frames.
# Original data is untouched; a new folder is created next to it.
# =====================================================================

import os
import shutil
import random
from pathlib import Path

# --- Config ---
SOURCE_DIR = '/content/drive/MyDrive/Colab Notebooks/Project_Orion/Labelled_Data/yolo_train_data'
SPLIT_DIR  = '/content/drive/MyDrive/Colab Notebooks/Project_Orion/Labelled_Data/yolo_train_val_split'
TRAIN_RATIO = 0.8
SEED = 42  # Reproducibility: same seed = same split every run

# --- Paths ---
src_images = Path(SOURCE_DIR) / 'images'
src_labels = Path(SOURCE_DIR) / 'labels'

# Verify source folders exist
assert src_images.exists(), f"Images folder not found: {src_images}"
assert src_labels.exists(), f"Labels folder not found: {src_labels}"

# --- Collect paired image/label files ---
image_extensions = {'.jpg', '.jpeg', '.png'}
image_files = [f for f in src_images.iterdir() if f.suffix.lower() in image_extensions]

# Only keep images that have a matching label file
paired_files = []
unmatched = []
for img in image_files:
    label_file = src_labels / f"{img.stem}.txt"
    if label_file.exists():
        paired_files.append(img.stem)  # stem = filename without extension
    else:
        unmatched.append(img.name)

print(f"Found {len(image_files)} images total")
print(f"Paired with labels: {len(paired_files)}")
if unmatched:
    print(f"⚠️ {len(unmatched)} images have no matching label — skipped:")
    for name in unmatched[:5]:
        print(f"   {name}")

# --- Reproducible shuffle ---
random.seed(SEED)
random.shuffle(paired_files)

# --- Split ---
split_idx = int(len(paired_files) * TRAIN_RATIO)
train_files = paired_files[:split_idx]
val_files   = paired_files[split_idx:]

print(f"\nSplit: {len(train_files)} train / {len(val_files)} val")

# --- Create target folder structure ---
for subset in ['train', 'val']:
    (Path(SPLIT_DIR) / 'images' / subset).mkdir(parents=True, exist_ok=True)
    (Path(SPLIT_DIR) / 'labels' / subset).mkdir(parents=True, exist_ok=True)

# --- Copy files ---
def copy_pair(stem, subset):
    # Find the actual image extension (could be .jpg, .png, etc.)
    for ext in image_extensions:
        img_src = src_images / f"{stem}{ext}"
        if img_src.exists():
            shutil.copy(img_src, Path(SPLIT_DIR) / 'images' / subset / img_src.name)
            break
    label_src = src_labels / f"{stem}.txt"
    shutil.copy(label_src, Path(SPLIT_DIR) / 'labels' / subset / label_src.name)

print("\nCopying files...")
for stem in train_files:
    copy_pair(stem, 'train')
for stem in val_files:
    copy_pair(stem, 'val')

print("Done.")

# --- Verify final counts ---
print("\n=== Final split counts ===")
for subset in ['train', 'val']:
    img_count = len(list((Path(SPLIT_DIR) / 'images' / subset).iterdir()))
    lbl_count = len(list((Path(SPLIT_DIR) / 'labels' / subset).iterdir()))
    print(f"{subset}: {img_count} images, {lbl_count} labels")
```

```python
from IPython.display import display, Image as IPImage
from collections import defaultdict
import glob, os, random
from pathlib import Path


VAL_DIR = '/content/drive/MyDrive/Colab Notebooks/Project_Orion/Labelled_Data/yolo_train_val_split/images/val'

all_val_images = sorted(glob.glob(f'{VAL_DIR}/*.jpg'))
print(f"Total val images available: {len(all_val_images)}")

NUM_TO_DISPLAY = 6
random.seed(42)  # reproducible selection
sample_images = random.sample(all_val_images, min(NUM_TO_DISPLAY, len(all_val_images)))

print(f"Selected {len(sample_images)} images\n")

summary = []

for i, image_path in enumerate(sample_images):

    # Run inference on this single image
    result = model.predict(
        source=image_path,
        conf=0.3,
        save=True,
        verbose=False,
        max_det=50
    )

    # Count detections
    class_counts = defaultdict(int)
    for box in result[0].boxes:
        class_name = model.names[int(box.cls[0])]
        class_counts[class_name] += 1

    car = class_counts.get('CAR', 0)
    gcs = class_counts.get('GCS', 0)
    ref = class_counts.get('REF', 0)
    total = car + gcs + ref

    summary.append({
        'image': os.path.basename(image_path),
        'CAR': car, 'GCS': gcs, 'REF': ref, 'total': total
    })


    saved_path = Path(result[0].save_dir) / os.path.basename(image_path)

    print(f"Image {i+1}: {os.path.basename(image_path)}")
    print(f"  CAR: {car}  |  GCS: {gcs}  |  REF: {ref}  |  Total: {total}")
    display(IPImage(str(saved_path), width=750))
    print()

# Summary table
print("\n" + "=" * 60)
print(f"{'Image':<40} {'CAR':>4} {'GCS':>4} {'REF':>4} {'Total':>6}")
print("-" * 60)
for s in summary:
    print(f"{s['image']:<40} {s['CAR']:>4} {s['GCS']:>4} {s['REF']:>4} {s['total']:>6}")
print("-" * 60)
totals = [sum(s[k] for s in summary) for k in ['CAR','GCS','REF','total']]
print(f"{'TOTAL':<40} {totals[0]:>4} {totals[1]:>4} {totals[2]:>4} {totals[3]:>6}")
```

###  The tracking code

```python
import cv2
import csv
from ultralytics import YOLO
from deep_sort_realtime.deepsort_tracker import DeepSort

# Load your trained YOLO model
model = YOLO(MODEL_PATH)

# Initialise DeepSORT
tracker = DeepSort(max_age=30)

# Class names — must match your training order
class_names = {0: 'CAR', 1: 'GCS', 2: 'REF'}

# Open video
cap = cv2.VideoCapture(VIDEO_PATH)
fps = cap.get(cv2.CAP_PROP_FPS)
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Setup output video writer
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(OUTPUT_VIDEO, fourcc, fps, (width, height))

# Setup CSV
csv_file = open(OUTPUT_CSV, 'w', newline='')
csv_writer = csv.writer(csv_file)
csv_writer.writerow(['frame', 'player_id', 'class', 'x_center', 'y_center', 'width', 'height', 'confidence'])

frame_number = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Run YOLO detection on current frame
    results = model(frame, verbose=False)[0]

    # Format detections for DeepSORT
    # DeepSORT expects: [ [x1, y1, w, h], confidence, class ]
    detections = []
    for box in results.boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        w = x2 - x1
        h = y2 - y1
        conf = float(box.conf[0])
        cls = int(box.cls[0])

        if conf > 0.4:  # confidence threshold — filter weak detections
            detections.append(([x1, y1, w, h], conf, cls))

    # Update DeepSORT tracker
    tracks = tracker.update_tracks(detections, frame=frame)

    # Loop through confirmed tracks
    for track in tracks:
        if not track.is_confirmed():
            continue

        track_id = track.track_id
        ltrb = track.to_ltrb()  # left, top, right, bottom
        x1, y1, x2, y2 = map(int, ltrb)
        x_center = (x1 + x2) // 2
        y_center = (y1 + y2) // 2
        w = x2 - x1
        h = y2 - y1
        cls = track.det_class
        class_label = class_names.get(cls, 'UNKNOWN')

        # Write to CSV
        csv_writer.writerow([frame_number, track_id, class_label, x_center, y_center, w, h, track.det_conf])

        # Draw on frame
        color = {'CAR': (255, 0, 0), 'GCS': (0, 255, 0), 'REF': (0, 0, 255)}.get(class_label, (255,255,255))
        cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
        cv2.putText(frame, f'{class_label} #{track_id}', (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)

    out.write(frame)
    frame_number += 1

    if frame_number % 100 == 0:
        print(f'Processed {frame_number} frames...')

# Cleanup
cap.release()
out.release()
csv_file.close()
print(f'Done! CSV saved to {OUTPUT_CSV}')
print(f'Video saved to {OUTPUT_VIDEO}')
```

### Remap new dataset classes to unified class list

```python
import os

# Current classes in new dataset
# 0: BALL, 1: CAT, 2: HAW, 3: POST, 4: REF

# Unified class list we want for combined dataset
# 0: CAR, 1: GCS, 2: HAW, 3: CAT, 4: REF, 5: BALL, 6: POST

# Mapping: old class number → new class number
remap = {
    0: 5,  # BALL → 5
    1: 3,  # CAT → 3
    2: 2,  # HAW → 2
    3: 6,  # POST → 6
    4: 4,  # REF → 4
}

labels_dir = '/content/dataset2/labels'
updated = 0

for txt_file in os.listdir(labels_dir):
    if not txt_file.endswith('.txt'):
        continue

    filepath = os.path.join(labels_dir, txt_file)

    with open(filepath, 'r') as f:
        lines = f.readlines()

    new_lines = []
    for line in lines:
        parts = line.strip().split()
        if len(parts) == 0:
            continue
        old_class = int(parts[0])
        new_class = remap[old_class]
        parts[0] = str(new_class)
        new_lines.append(' '.join(parts))

    with open(filepath, 'w') as f:
        f.write('\n'.join(new_lines))

    updated += 1

print(f"Remapped {updated} label files successfully!")
```

### Remap old dataset REF class

```python
import os
import shutil

old_labels_dir = '/content/drive/MyDrive/Colab Notebooks/Project_Orion/Labelled_Data/yolo_train_val_split'

old_remap = {
    0: 0,  # CAR stays 0
    1: 1,  # GCS stays 1
    2: 4,  # REF → 4
}

os.makedirs('/content/old_dataset_remapped/train/images', exist_ok=True)
os.makedirs('/content/old_dataset_remapped/train/labels', exist_ok=True)
os.makedirs('/content/old_dataset_remapped/val/images', exist_ok=True)
os.makedirs('/content/old_dataset_remapped/val/labels', exist_ok=True)

for split in ['train', 'val']:
    # Copy images
    img_src = f'{old_labels_dir}/images/{split}'
    img_dst = f'/content/old_dataset_remapped/{split}/images'
    for img in os.listdir(img_src):
        shutil.copy(os.path.join(img_src, img), os.path.join(img_dst, img))

    # Remap labels
    lbl_src = f'{old_labels_dir}/labels/{split}'
    lbl_dst = f'/content/old_dataset_remapped/{split}/labels'
    for txt_file in os.listdir(lbl_src):
        if not txt_file.endswith('.txt'):
            continue
        with open(os.path.join(lbl_src, txt_file), 'r') as f:
            lines = f.readlines()
        new_lines = []
        for line in lines:
            parts = line.strip().split()
            if len(parts) == 0:
                continue
            old_class = int(parts[0])
            new_class = old_remap[old_class]
            parts[0] = str(new_class)
            new_lines.append(' '.join(parts))
        with open(os.path.join(lbl_dst, txt_file), 'w') as f:
            f.write('\n'.join(new_lines))

print("Old dataset remapped successfully!")
```

```python
from ultralytics import YOLO
import glob
import cv2
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
import numpy as np

# Load your new unified model
model = YOLO('/content/drive/MyDrive/Colab Notebooks/Project_Orion/AFL_Model/best_unified.pt')

# Class names and colours
class_names = ['CAR', 'GCS', 'HAW', 'CAT', 'REF', 'BALL', 'POST']
class_colors = {
    'CAR': (0, 0, 255),
    'GCS': (255, 165, 0),
    'HAW': (255, 100, 0),
    'CAT': (26, 62, 143),
    'REF': (255, 215, 0),
    'BALL': (255, 0, 0),
    'POST': (0, 255, 0)
}

# Grab 4 sample images from val set
test_images = glob.glob('/content/combined_dataset/val/images/*.jpg')[:4]
print(f"Testing on {len(test_images)} images")

# Run inference
results = model.predict(
    source=test_images,
    conf=0.3,
    verbose=False
)

# Display all 4 images in a grid
fig, axes = plt.subplots(2, 2, figsize=(20, 14))
axes = axes.flatten()

for i, (result, img_path) in enumerate(zip(results, test_images)):
    # Read image
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Draw each detection
    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        class_name = class_names[class_id]
        color = class_colors[class_name]

        # Get box coordinates (x1, y1, x2, y2)
        x1, y1, x2, y2 = map(int, box.xyxy[0])

        # Draw rectangle
        cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)

        # Draw label with confidence score
        label = f"{class_name} {confidence:.2f}"
        label_size, _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 2)
        cv2.rectangle(img, (x1, y1 - label_size[1] - 8), (x1 + label_size[0], y1), color, -1)
        cv2.putText(img, label, (x1, y1 - 4), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)

    # Show image
    axes[i].imshow(img)
    axes[i].set_title(f"Image {i+1}: {img_path.split('/')[-1]}", fontsize=10)
    axes[i].axis('off')

    # Print detections to console as well
    print(f"\nImage {i+1}: {img_path.split('/')[-1]}")
    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        class_name = class_names[class_id]
        print(f"  {class_name}: {confidence:.2f}")

plt.tight_layout()
plt.savefig('/content/unified_model_test.jpg', dpi=150, bbox_inches='tight')
plt.show()
print("\nSaved output image to /content/unified_model_test.jpg")
```

### Tracking Code

### paths

```python
MODEL_PATH = '/content/drive/MyDrive/Colab Notebooks/Project_Orion/AFL_Model/best_unified.pt'
VIDEO_PATH = '/content/drive/MyDrive/Colab Notebooks/Project_Orion/haw_cat_match.mp4'
OUTPUT_VIDEO = '/content/drive/MyDrive/Colab Notebooks/Project_Orion/haw_cat_tracked.mp4'
OUTPUT_CSV = '/content/drive/MyDrive/Colab Notebooks/Project_Orion/haw_cat_tracking.csv'
```

```python
import cv2
import csv
from ultralytics import YOLO
from deep_sort_realtime.deepsort_tracker import DeepSort

model = YOLO(MODEL_PATH)
tracker = DeepSort(max_age=30)

# Updated for unified model
class_names = {0: 'CAR', 1: 'GCS', 2: 'HAW', 3: 'CAT', 4: 'REF', 5: 'BALL', 6: 'POST'}
class_colors = {
    'CAR': (255, 0, 0),
    'GCS': (0, 165, 255),
    'HAW': (0, 100, 255),
    'CAT': (143, 62, 26),
    'REF': (0, 215, 255),
    'BALL': (0, 0, 255),
    'POST': (0, 255, 0)
}

cap = cv2.VideoCapture(VIDEO_PATH)
fps = cap.get(cv2.CAP_PROP_FPS)
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(OUTPUT_VIDEO, fourcc, fps, (width, height))

csv_file = open(OUTPUT_CSV, 'w', newline='')
csv_writer = csv.writer(csv_file)
csv_writer.writerow(['frame', 'player_id', 'class', 'x_center', 'y_center', 'width', 'height', 'confidence'])

frame_number = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame, verbose=False)[0]

    detections = []
    for box in results.boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        w = x2 - x1
        h = y2 - y1
        conf = float(box.conf[0])
        cls = int(box.cls[0])
        if conf > 0.4:
            detections.append(([x1, y1, w, h], conf, cls))

    tracks = tracker.update_tracks(detections, frame=frame)

    for track in tracks:
        if not track.is_confirmed():
            continue

        track_id = track.track_id
        ltrb = track.to_ltrb()
        x1, y1, x2, y2 = map(int, ltrb)
        x_center = (x1 + x2) // 2
        y_center = (y1 + y2) // 2
        w = x2 - x1
        h = y2 - y1
        cls = track.det_class
        class_label = class_names.get(cls, 'UNKNOWN')
        color = class_colors.get(class_label, (255, 255, 255))

        csv_writer.writerow([frame_number, track_id, class_label, x_center, y_center, w, h, track.det_conf])

        cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
        cv2.putText(frame, f'{class_label} #{track_id}', (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)

    out.write(frame)
    frame_number += 1

    if frame_number % 100 == 0:
        print(f'Processed {frame_number} frames...')

cap.release()
out.release()
csv_file.close()
print(f'Done! CSV saved to {OUTPUT_CSV}')
print(f'Video saved to {OUTPUT_VIDEO}')
```
